```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.mnist.load_data()
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
x_train[2]
```

```python
plt.matshow(x_train[2])
```

```python
x_train.shape
```

```python
x_train=x_train/255
x_test=x_test/255
```

```python
x_train_flattened=x_train.reshape(len(x_train),28*28)
x_test_flattened=x_test.reshape(len(x_test),28*28)
```

```python
x_train_flattened.shape
```

```python
x_test_flattened.shape
```

```python
x_train_flattened[0]
```

```python
model=keras.Sequential([
    keras.layers.Dense(10,input_shape=(784,),activation='sigmoid')
])
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model.fit(x_train_flattened,y_train,epochs=5)
```

```python
model.evaluate(x_test_flattened,y_test)
```

```python
predict=model.predict(x_test_flattened)
```

```python
predict[0]
```

```python
np.argmax(predict[1])
```

```python
y_predicted=[np.argmax(i) for i in predict]
y_predicted[:5]
```

```python
y_test[:5]
```

```python
cm=tf.math.confusion_matrix(labels=y_test,predictions=y_predicted)
cm
```

```python
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model=keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(100,activation='relu'),
    keras.layers.Dense(10,activation='sigmoid')
])
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model.fit(x_train,y_train,epochs=5)
```

```python
from tensorflow.keras import datasets,layers,models
```

```python
cnn=models.Sequential([
    layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu",input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu"),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(100,activation="relu"),
    layers.Dense(10,activation="softmax")
])
cnn.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
cnn.fit(x_train,y_train,epochs=10)
```

```python
cnn2=models.Sequential([
    layers.Conv2D(filters=32,kernel_size=(3,3),padding="same",activation="relu",input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(filters=32,kernel_size=(3,3),activation="relu"),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(100,activation="relu"),
    layers.Dense(10,activation="softmax")
])
cnn2.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
cnn2.fit(x_train,y_train,epochs=10)
```
